## Tháp Hà Nội - Trực quan hóa các bước đi

```python
def tower_of_hanoi(n, source, auxiliary, destination, poles, n_disks, move_history):
    if n == 1:
        disk = poles[source].pop(0)
        poles[destination].insert(0, disk)
        move_history[0] += 1
        print(f"Bước {move_history[0]}: Di chuyển đĩa {disk} từ cọc {chr(ord('A') + source)} đến cọc {chr(ord('A') + destination)}")
        return

    tower_of_hanoi(n - 1, source, destination, auxiliary, poles, n_disks, move_history)

    disk = poles[source].pop(0)
    poles[destination].insert(0, disk)
    move_history[0] += 1
    print(f"Bước {move_history[0]}: Di chuyển đĩa {disk} từ cọc {chr(ord('A') + source)} đến cọc {chr(ord('A') + destination)}")

    tower_of_hanoi(n - 1, auxiliary, source, destination, poles, n_disks, move_history)

# --- Chương trình chính ---
n_disks = 3 # Số đĩa bạn muốn giải (có thể thay đổi)

poles = [
    [i for i in range(1, n_disks + 1)], # Cọc A (nguồn) chứa tất cả đĩa
    [], # Cọc B (trung gian) rỗng
    []  # Cọc C (đích) rỗng
]

move_history = [0]

print(f"Bắt đầu giải Tháp Hà Nội với {n_disks} đĩa.")

tower_of_hanoi(n_disks, 0, 1, 2, poles, n_disks, move_history)

print(f"Hoàn thành sau {move_history[0]} bước di chuyển.")
print(f"Trạng thái cuối cùng của các cọc: A:{poles[0]}, B:{poles[1]}, C:{poles[2]}")
```

```python
from collections import defaultdict, deque

def input_graph():
    graph = defaultdict(list)
    # Nhập số đỉnh và số cạnh
    n, m = map(int, input("Nhập số đỉnh và số cạnh (ví dụ: 5 4): ").split())

    print(f"Nhập {m} cạnh (mỗi cạnh gồm 2 đỉnh u v, ví dụ: 1 2):")
    for _ in range(m):
        u, v = map(int, input().split())
        # Đồ thị vô hướng, các cạnh có vai trò như nhau
        graph[u].append(v)
        graph[v].append(u)

    # Sắp xếp lại danh sách kề để thứ tự duyệt các đỉnh con được đồng nhất
    for node in graph:
        graph[node].sort()

    return graph

def print_graph(graph):
    print("\n--- DANH SÁCH KỀ CỦA ĐỒ THỊ ---")
    for node in sorted(graph.keys()):
        print(f"Đỉnh {node} kết nối với: {graph[node]}")

def bfs(graph, start_node):
    visited = set()
    queue = deque([start_node])
    visited.add(start_node)

    result = []
    while queue:
        vertex = queue.popleft()
        result.append(str(vertex))

        # Duyệt qua các đỉnh kề của đỉnh hiện tại
        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    print(f"Kết quả duyệt BFS từ đỉnh {start_node}: " + " -> ".join(result))

def dfs(graph, start_node):
    visited = set()
    result = []

    # Định nghĩa hàm đệ quy bên trong để duyệt DFS
    def dfs_recursive(vertex):
        visited.add(vertex)
        result.append(str(vertex))
        for neighbor in graph[vertex]:
            if neighbor not in visited:
                dfs_recursive(neighbor)

    dfs_recursive(start_node)
    print(f"Kết quả duyệt DFS từ đỉnh {start_node}: " + " -> ".join(result))

# --- CHƯƠNG TRÌNH CHÍNH ---
if __name__ == "__main__":
    # 1. Nhập dữ liệu đồ thị
    my_graph = input_graph()

    # 2. In cấu trúc đồ thị ra màn hình
    print_graph(my_graph)

    # Chọn đỉnh bắt đầu để duyệt (lấy đỉnh đầu tiên xuất hiện trong danh sách)
    if my_graph:
        start = min(my_graph.keys())
        print("\n--- KẾT QUẢ DUYỆT ĐỒ THỊ ---")
        # 3. Duyệt BFS
        bfs(my_graph, start)
        # 4. Duyệt DFS
        dfs(my_graph, start)
    else:
        print("Đồ thị rỗng!")
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# ==========================================
# 1. KHỞI TẠO CÂY ĐỒ THỊ (Cây nhị phân mẫu)
# ==========================================
# Tạo một đồ thị có hướng để biểu diễn cấu trúc cây
G = nx.DiGraph()

# Định nghĩa các cạnh của cây (Gốc là A)
edges = [
    ('A', 'B'), ('A', 'C'),
    ('B', 'D'), ('B', 'E'),
    ('C', 'F'), ('C', 'G'),
    ('D', 'H'), ('D', 'I')
]
G.add_edges_from(edges)

# Tọa độ cố định để vẽ đồ thị thành hình cây phân cấp (tùy chỉnh cho đẹp mắt)
pos = {
    'A': (0, 3),
    'B': (-2, 2), 'C': (2, 2),
    'D': (-3, 1), 'E': (-1, 1), 'F': (1, 1), 'G': (3, 1),
    'H': (-3.5, 0), 'I': (-2.5, 0)
}

# ==========================================
# 2. THUẬT TOÁN DUYỆT CÂY & ĐÁNH DẤU THỨ TỰ
# ==========================================
# Lấy danh sách các nút theo thứ tự duyệt BFS và DFS của NetworkX
bfs_order = list(nx.bfs_tree(G, source='A').nodes())
dfs_order = list(nx.dfs_tree(G, source='A').nodes())

print("Thứ tự duyệt Chiều rộng (BFS):", " -> ".join(bfs_order))
print("Thứ tự duyệt Chiều sâu (DFS):", " -> ".join(dfs_order))

# ==========================================
# 3. HÀM TRỰC QUAN HÓA (VẼ ĐỒ THỊ)
# ==========================================
def draw_tree_search(order_list, title):
    # Tạo một bảng màu dựa trên thứ tự duyệt (Nút duyệt đầu tiên sẽ đậm nhất)
    # Tạo dict lưu vị trí (thứ tự) xuất hiện của nút trong danh sách duyệt
    order_map = {node: i for i, node in enumerate(order_list)}

    # Gán màu cho từng nút (giá trị từ 0.2 đến 1.0 để làm độ đậm nhạt)
    node_colors = [order_map[node] for node in G.nodes()]

    plt.figure(figsize=(8, 6))

    # Vẽ các cạnh và nút
    nodes = nx.draw_networkx_nodes(
        G, pos,
        node_color=node_colors,
        cmap=plt.cm.Blues, # Sử dụng palette màu Xanh dương
        node_size=800,
        edgecolors='black'
    )
    nx.draw_networkx_edges(G, pos, arrowstyle='->', arrowsize=15, edge_color='gray')

    # Vẽ nhãn chữ (Tên nút) kèm theo thứ tự duyệt nhỏ bên cạnh
    labels = {node: f"{node}\n(#{order_map[node] + 1})" for node in G.nodes()}
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')

    # Thêm thanh hiển thị mức độ tiến trình duyệt (Colorbar)
    cbar = plt.colorbar(nodes, ax=plt.gca(), orientation='horizontal', pad=0.05)
    cbar.set_label('Thứ tự tiến trình duyệt (Nhạt dần = Duyệt sau cùng)', fontsize=10)
    cbar.set_ticks([]) # Ẩn số của colorbar để nhìn gọn hơn

    plt.title(title, fontsize=14, fontweight='bold', pad=20)
    plt.axis('off')

# ==========================================
# 4. CHẠY VẼ MINH HỌA
# ==========================================
# Vẽ đồ thị mô phỏng Duyệt chiều rộng (BFS)
draw_tree_search(bfs_order, "Trực quan hóa Duyệt theo Chiều rộng (BFS)")

# Vẽ đồ thị mô phỏng Duyệt chiều sâu (DFS)
draw_tree_search(dfs_order, "Trực quan hóa Duyệt theo Chiều sâu (DFS)")

# Hiển thị cả 2 cửa sổ đồ thị
plt.show()
```
