```python
import sympy
from sympy import to_cnf
from sympy.parsing.sympy_parser import parse_expr

def string_to_cnf(expression_str):
    """
    Hàm nhận vào chuỗi logic, chuyển đổi các ký hiệu phổ thông
    thành cú pháp SymPy và trả về chuỗi kết quả dạng CNF.
    """
    # Chuẩn hóa các ký hiệu logic phổ biến về định dạng của SymPy
    # SymPy sử dụng: & (AND), | (OR), ~ (NOT), >> (IMPLIES)
    expr_normalized = (expression_str
                       .replace("AND", "&")
                       .replace("and", "&")
                       .replace("OR", "|")
                       .replace("or", "|")
                       .replace("NOT", "~")
                       .replace("not", "~")
                       .replace("=>", ">>")
                       .replace("implies", ">>"))

    try:
        # Biến đổi chuỗi thành biểu thức của SymPy
        expr = parse_expr(expr_normalized, evaluate=False)
        # Chuyển đổi biểu thức sang dạng hội chuẩn CNF
        cnf_expr = to_cnf(expr)
        return cnf_expr
    except Exception as e:
        return f"Lỗi cú pháp biểu thức: {e}"

# --- CHƯƠNG TRÌNH CHẠY THỬ VỚI CÁC VÍ DỤ ---
if __name__ == "__main__":
    # Danh sách các ví dụ mệnh đề logic phổ biến
    examples = [
        "p implies q", # Kéo theo: p -> q (Kết quả CNF: ~p | q)
        "not (p or q)", # Luật De Morgan: ~(p v q) (Kết quả CNF: ~p & ~q)
        "(p and q) or r", # Luật phân phối (Kết quả CNF: (p | r) & (q | r))
        "(p implies q) and (q implies r)", # Hệ quả bắc cầu
        "p or (q and (r or not s))" # Biểu thức phức tạp hơn
    ]

    print("=== KẾT QUẢ CHUYỂN ĐỔI CÁC VÍ DỤ CÓ SẴN ===")
    for idx, ex in enumerate(examples, 1):
        print(f"Ví dụ {idx}: {ex}")
        print(f" -> Dạng CNF: {string_to_cnf(ex)}")
        print("-" * 40)

    # --- PHẦN NHẬP DỮ LIỆU TỪ BÀN PHÍM ---
    print("\n=== BẠN TỰ NHẬP MỆNH ĐỀ ĐỂ KIỂM TRA ===")
    print("Ký hiệu hợp lệ: and (&&), or (||), not (~), implies (=>)")
    user_input = input("Nhập mệnh đề logic của bạn: ")

    result = string_to_cnf(user_input)
    print(f" -> Kết quả CNF: {result}")
```

```python
import itertools
def implies(a, b):
    """Phép kéo theo: a -> b tương đương not a or b"""
    return (not a) or b


# Danh sách các mệnh đề và ý nghĩa
meanings = {
    "P1": "Nam là chuyên gia",
    "P2": "Nam là người cá biệt",
    "P3": "Nam có nhiều báo cáo có tiếng",
    "P4": "Nam được đồng nghiệp tin cậy",
    "P5": "Hộp thư của Nam có nhiều thư",
    "P6": "Nam được bạn bè tôn trọng",
}

variables = ["P1", "P2", "P3", "P4", "P5", "P6"]
solutions = []

# Duyệt qua tất cả các tổ hợp chân trị (True/False) của 6 biến
for values in itertools.product([True, False], repeat=6):
    P1, P2, P3, P4, P5, P6 = values

    # 5 điều kiện đã cho:
    c1 = (P1 and not P2) or (not P1 and P2) # Câu 1
    c2 = implies(P1, P3 and P4) # Câu 2
    c3 = implies(P3, P5) # Câu 3
    c4 = implies(P2, not P6) # Câu 4
    c5 = not P5 # Câu 5

    # Kiểm tra xem tất cả các câu có đồng thời đúng hay không
    if c1 and c2 and c3 and c4 and c5:
        solutions.append(dict(zip(variables, values)))

# In kết quả
print(f"Tìm thấy {len(solutions)} mô hình thỏa mãn:\n")
for idx, sol in enumerate(solutions, 1):
    print(f"--- Mô hình {idx} ---")
    for var, val in sol.items():
        print(f" {var} ({meanings[var]}): {val}")
    print()
```
